```python
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
df_train = pd.read_csv('../data/train.csv')
```

```python
df_train['target'].value_counts().apply(lambda x: x / len(df_train))
```

```python
df_train['target'].value_counts(normalize=True).sort_index().plot(
    kind='bar',
    figsize=(6, 4)
)

plt.title('Распределение классов')
plt.xlabel('Класс')
plt.ylabel('Доля')
plt.xticks(rotation=0)
plt.grid(axis='y', alpha=0.3)
plt.show()
```
